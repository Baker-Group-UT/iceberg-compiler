```python
config = PipelineConfig(
    sim_config=ScaledSimulator(alpha=1, repetitions=500).with_seed(6),
    verbose=True,
    save=False,
    seed=4,
)
print(pack(circuit=pcirc))

res = run_full_scan(pcirc, config)
```

```python
print(len(res.dataframe.df[res.dataframe.df["fidelity"] < 0.80]["permutation"]))
for permutation in res.dataframe.df[res.dataframe.df["fidelity"] < 0.80]["permutation"]:
    print(translate_permutation(pcirc, permutation=permutation, config=config).circuit.draw(fold=1000))
```
