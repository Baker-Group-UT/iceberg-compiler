```python

import sys

from pipeline import PipelineConfig, generate_permutation_set, run_full_scan, run_specific_scan
from simulator import SimulatorConfig
from simulator.config_presets import StandardSimulator
sys.path.append("../../../")

from qiskit import QuantumCircuit

from packer.packer import HeuristicPacker


qc= QuantumCircuit(4)
qc.h(0)
qc.h(3)
qc.x(1)
qc.cx(0,1)
qc.cx(3,2)
qc.h(2)
qc.z(2)
qc.measure_all()

packer = HeuristicPacker()
packing2 = packer.pack(qc, max_patches=2)
permutation2 = packer.packing_to_permutation(packing2.qubit_to_patch,)
packing4 = packer.pack(qc, max_patches=4)
permutation4 = packer.packing_to_permutation(packing4.qubit_to_patch)
print(packing2)
print(permutation2)
print(packing4)
print(permutation4)

config = PipelineConfig(
    sim_config=StandardSimulator(repetitions=100_000),
    verbose=False,
)

perm_set = generate_permutation_set(qc, config)
perm_set.append(permutation2)
perm_set.append(permutation4)

res = run_specific_scan(qc, perm_set, config)
```

```python

import os
from typing import Optional, Sequence, Tuple

import numpy as np
import matplotlib.pyplot as plt

if res.dataframe is None:
    raise ValueError("No dataframe attached to run results")

df = res.dataframe.df.copy()
df = df.dropna(subset=["codes_used"])
df["codes_used"] = df["codes_used"].astype(int)


def perm_to_key(perm):
    if perm is None:
        return None
    return tuple(tuple(p) for p in perm)

highlight_perms = {
    "Permutation (2 patches)": perm_to_key(permutation2),
    "Permutation (4 patches)": perm_to_key(permutation4),
}

highlight_colors = {
    "Permutation (2 patches)": "#E4572E",
    "Permutation (4 patches)": "#1D6996",
}


def _get_first_present_col(df, candidates: Sequence[str]) -> str:
    for col in candidates:
        if col in df.columns:
            return col
    raise KeyError(f"None of the candidate columns found. Looked for: {candidates}")


def plot_violin_by_codes(
    df,
    value_col: str,
    group_col: str = "codes_used",
    title: Optional[str] = None,
) -> Tuple[plt.Figure, plt.Axes]:

    missing = [c for c in (value_col, group_col) if c not in df.columns]
    if missing:
        raise KeyError(f"DataFrame missing required columns: {missing}")

    clean = df[[group_col, value_col]].dropna()
    if clean.empty:
        raise ValueError("No data available after dropping NA values.")

    groups: Sequence[int] = sorted(clean[group_col].unique())
    datasets = [clean.loc[clean[group_col] == g, value_col].to_numpy() for g in groups]
    positions = np.arange(len(groups)) + 1

    fig, ax = plt.subplots(figsize=(8, 5))
    parts = ax.violinplot(
        datasets,
        positions=positions,
        showmeans=True,
        showmedians=False,
        showextrema=True,
    )

    for body in parts.get("bodies", []):
        body.set_facecolor("#BFD3F8")
        body.set_edgecolor("#4C78A8")
        body.set_alpha(0.65)

    for key in ("cmeans", "cbars", "cmins", "cmaxes"):
        artist = parts.get(key)
        if artist is not None:
            artist.set_edgecolor("#1E1E1E")
            artist.set_linewidth(0.8)

    rng = np.random.default_rng(0)
    for idx, data in enumerate(datasets, start=1):
        if len(data) == 0:
            continue
        x = np.full(len(data), idx, dtype=float) + rng.normal(0, 0.03, size=len(data))
        ax.plot(x, data, linestyle="", marker="o", markersize=3, alpha=0.6, color="#305F72")

    ax.set_xlabel("# of codes used")
    ax.set_ylabel(value_col)
    ax.set_title(title or f"{value_col} by # codes used (Violin)")
    ax.set_xticks(positions)
    ax.set_xticklabels([str(g) for g in groups])
    ax.grid(True, axis="y", linestyle="--", linewidth=0.5, alpha=0.5)
    fig.tight_layout()

    return fig, ax


tvd_col = _get_first_present_col(
    df,
    candidates=["logical_total_variation_distance", "TVD", "tvd"],
)
sel_col = _get_first_present_col(
    df,
    candidates=["logical_selection_rate", "selection_rate", "selection rate"],
)

metrics = [
    (tvd_col, "Logical Total Variation Distance"),
    (sel_col, "Logical Selection Rate"),
]

for metric, y_label in metrics:
    clean_metric = df[["codes_used", metric]].dropna()
    if clean_metric.empty:
        continue
    groups = sorted(clean_metric["codes_used"].unique())
    group_positions = {g: idx + 1 for idx, g in enumerate(groups)}

    fig, ax = plot_violin_by_codes(
        df=df,
        value_col=metric,
        group_col="codes_used",
        title=f"{y_label} by Codes Used",
    )

    for label, perm_key in highlight_perms.items():
        matching_rows = df[(df["perm_key"] == perm_key) & df[metric].notna()]
        if matching_rows.empty:
            continue

        x_positions = matching_rows["codes_used"].map(group_positions)
        scatter = ax.scatter(
            x_positions,
            matching_rows[metric],
            s=90,
            label=label,
            zorder=6,
            marker="o",
            edgecolors="black",
            linewidths=0.8,
            facecolors=highlight_colors[label],
        )

        for _, row in matching_rows.iterrows():
            ax.annotate(
                label,
                (group_positions[row["codes_used"]], row[metric]),
                xytext=(6, 6),
                textcoords="offset points",
                fontsize=9,
                color=highlight_colors[label],
                fontweight="bold",
            )

    handles, labels = ax.get_legend_handles_labels()
    if handles:
        ax.legend(title="Highlighted permutations", loc="best")
    plt.show()
```
